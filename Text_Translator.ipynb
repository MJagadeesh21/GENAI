```python
import torch
from transformers import MarianMTModel, MarianTokenizer

# Detect GPU and move model to device
device = "cuda" if torch.cuda.is_available() else "cpu"

# Load the English-to-German model
model_name = "Helsinki-NLP/opus-mt-en-de"
tokenizer = MarianTokenizer.from_pretrained(model_name)
model = MarianMTModel.from_pretrained(model_name).to(device)

# Function to translate text
def translate(text):
    text = text.strip()  # Remove extra spaces
    if not text:
        return "Error: Empty input!"

    # Tokenize and convert input to tensor
    inputs = tokenizer(text, return_tensors="pt", padding=True, truncation=True).to(device)

    # Generate translation
    translated = model.generate(**inputs)

    # Decode output
    return tokenizer.batch_decode(translated, skip_special_tokens=True)[0]

# Interactive translation loop
while True:
    text = input("Enter text in English (or type 'exit' to stop): ")
    if text.lower() == "exit":
        print("Translation stopped.")
        break
    translated_text = translate(text)
    print(f"German Translation: {translated_text}\n")
```

```python
import torch
from transformers import MarianMTModel, MarianTokenizer

# Detect GPU and move model to device
device = "cuda" if torch.cuda.is_available() else "cpu"

# Load the German-to-English model
model_name = "Helsinki-NLP/opus-mt-de-en"  # Changed model for DE → EN
tokenizer = MarianTokenizer.from_pretrained(model_name)
model = MarianMTModel.from_pretrained(model_name).to(device)

# Function to translate text
def translate(text):
    text = text.strip()  # Remove extra spaces
    if not text:
        return "Error: Empty input!"

    # Tokenize and convert input to tensor
    inputs = tokenizer(text, return_tensors="pt", padding=True, truncation=True).to(device)

    # Generate translation
    translated = model.generate(**inputs)

    # Decode output
    return tokenizer.batch_decode(translated, skip_special_tokens=True)[0]

# Interactive translation loop
while True:
    text = input("Enter text in German (or type 'exit' to stop): ")  # Changed prompt to German
    if text.lower() == "exit":
        print("Translation stopped.")
        break
    translated_text = translate(text)
    print(f"English Translation: {translated_text}\n")  # Changed output format
```
