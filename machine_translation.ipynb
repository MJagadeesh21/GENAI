```python
from transformers import pipeline

# Load the translation pipeline
translator = pipeline("translation", model="t5-base")

# Define the translation function
def translate(text):
    result = translator(text)
    return result[0]['translation_text']
```

```python
import gradio as gr

# Define the Gradio interface
demo = gr.Interface(
    fn=translate,  # Function to execute
    inputs=gr.Textbox(label="Input Text"),  # Input component
    outputs=gr.Textbox(label="Translated Text"),  # Output component
    title="Machine Translation",
    description="Enter text in English to translate it into another language."
)

# Launch the app
demo.launch()
```

```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

model = AutoModelForSeq2SeqLM.from_pretrained("facebook/nllb-200-distilled-600M")
tokenizer = AutoTokenizer.from_pretrained("facebook/nllb-200-distilled-600M")

LANGS = ["eng_Latn", "fra_Latn", "spa_Latn"]  # Example languages

def translate_multi(text, src_lang, tgt_lang):
    translation_pipeline = pipeline(
        "translation", model=model, tokenizer=tokenizer,
        src_lang=src_lang, tgt_lang=tgt_lang
    )
    result = translation_pipeline(text)
    return result[0]['translation_text']

demo = gr.Interface(
    fn=translate_multi,
    inputs=[
        gr.Textbox(label="Text"),
        gr.Dropdown(label="Source Language", choices=LANGS),
        gr.Dropdown(label="Target Language", choices=LANGS),
    ],
    outputs=gr.Textbox(label="Translated Text"),
    title="Multi-Language Translator"
)
demo.launch()
```
