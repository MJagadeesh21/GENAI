```python
import openai

# Set your API key
openai.api_key = "your_openai_api_key"

# Define the Q&A function
def ask_question(question):
    response = openai.ChatCompletion.create(
        model="gpt-3.5-turbo",  # Use a model like GPT-3.5 or GPT-4
        messages=[{"role": "user", "content": question}]
    )
    return response['choices'][0]['message']['content']
```

```python
from transformers import pipeline

# Load the Q&A pipeline
qa_pipeline = pipeline("question-answering", model="deepset/roberta-base-squad2")

# Define the Q&A function
def ask_question_hf(question, context):
    result = qa_pipeline(question=question, context=context)
    return result['answer']
```

```python
import gradio as gr

# Define the Gradio interface
interface = gr.Interface(
    fn=ask_question,  # Function to call
    inputs=gr.Textbox(label="Your Question"),  # Input field
    outputs=gr.Textbox(label="Answer"),  # Output field
    title="Q&A Model",
    description="Ask any question, and the model will provide an answer."
)

# Launch the app
interface.launch()
```
