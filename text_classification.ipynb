```python
from transformers import pipeline

# Load a pre-trained text classification pipeline
classifier = pipeline("text-classification", model="distilbert-base-uncased-finetuned-sst-2-english")
```

```python
def classify_text(text):
    result = classifier(text)
    label = result[0]['label']
    confidence = result[0]['score']
    return f"Label: {label}, Confidence: {confidence:.2f}"
```

```python
import gradio as gr

# Create the Gradio interface
demo = gr.Interface(
    fn=classify_text,
    inputs=gr.Textbox(lines=2, placeholder="Enter text to classify..."),
    outputs="text",
    title="Text Classification App",
    description="Classify text as Positive or Negative sentiment.",
    examples=[
        ["I love this product!"],
        ["This is the worst experience ever."],
        ["It's okay, nothing special."]
    ]
)

# Launch the app
demo.launch()
```
